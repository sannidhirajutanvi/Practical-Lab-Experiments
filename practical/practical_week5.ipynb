```python
import pandas as pd

df = pd.read_csv("/content/placement_predict_50k Dataset (1).csv")

# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

# Save the filled dataset
df.to_csv("placementfilledcat.csv", index=False)

# Check missing values
print(df.isnull().sum())
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Select only two numerical features
X = df[['CodingTestScore', 'AptitudeTestScore']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(
    X_scaled,
    columns=['CodingTestScore', 'AptitudeTestScore']
)

# Display results
print("Original Data")
print(X.head())

print("\nAfter Standard Scaling")
print(scaled_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Select only two numerical features
X = df[['CGPA', 'AptitudeTestScore']]

# Apply Min-Max Scaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(
    X_scaled,
    columns=['CGPA', 'AptitudeTestScore']
)

# Display results
print("Original Data")
print(X.head())

print("\nAfter Min-Max Scaling")
print(scaled_df.head())
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/placement_predict_50k Dataset (1).csv")

# Fill missing numerical values with column mean
df.fillna(df.mean(numeric_only=True), inplace=True)

# Save the updated dataset to a new CSV file
df.to_csv("placementfilled.csv", index=False)

# Check if any missing values remain
print(df.isnull().sum())

print("\nNew dataset saved as 'placementfilled.csv'")
```
